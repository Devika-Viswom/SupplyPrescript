# Sprint 5 - Feature Engineering

## Objective

Create new business features that improve the predictive power of the machine learning model.

```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/processed/cleaned_logistics_dataset.csv")

df.head()
```

```python
df["Cost_Per_KM"] = df["Fuel_Price_Index"] / df["Distance_km"]
```

```python
df["Risk_LeadTime"] = (
    df["Geopolitical_Risk_Score"] *
    df["Lead_Time_Days"]
)
```

```python
df["Weight_Per_KM"] = (
    df["Weight_MT"] /
    df["Distance_km"]
)
```

```python
df["Risk_Reliability"] = (
    df["Geopolitical_Risk_Score"] *
    (1 - df["Carrier_Reliability_Score"])
)
```

```python
df["High_Risk"] = (
    df["Geopolitical_Risk_Score"] >= 7
).astype(int)
```

```python
df.head()
```

```python
import os

os.makedirs("../data/features", exist_ok=True)
```

```python
df.to_csv(
    "../data/features/feature_dataset.csv",
    index=False
)
```
