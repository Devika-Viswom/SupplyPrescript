```python
import pandas as pd

from connection import conn
```

```python
df=pd.read_sql("SELECT * FROM raw1", con=conn)
```

```python
df_clean=df.copy()
```

```python
df_clean["Date"] = pd.to_datetime(df_clean["Date"])
```

```python
df_clean.info()
```

```python
df_clean["Year"] = df_clean["Date"].dt.year
df_clean["Month"] = df_clean["Date"].dt.month
df_clean["Quarter"] = df_clean["Date"].dt.quarter
```

```python
df_clean["Route"] = (
    df_clean["Origin_Port"]
    + "_to_" +
    df_clean["Destination_Port"]
)
```

```python
df_clean["Reliability_Level"] = pd.cut(
    df_clean["Carrier_Reliability_Score"],
    bins=[0,0.65,0.85,1],
    labels=["Low","Medium","High"]
)
```

```python
df_clean["Risk_Level"] = pd.cut(
    df_clean["Geopolitical_Risk_Score"],
    bins=[0,3,7,10],
    labels=["Low","Medium","High"],
    include_lowest=True
)
```
