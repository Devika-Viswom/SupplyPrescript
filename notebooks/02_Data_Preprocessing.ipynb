## Objective

The objective of this notebook is to clean and transform the logistics dataset into a machine-learning-ready dataset.

The preprocessing steps include:

- Loading the dataset
- Converting data types
- Feature engineering
- Encoding categorical variables
- Saving the processed dataset/

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/logistics_dataset.csv")

df.head()
```

```python
df_processed = df.copy()
```

```python
df_processed.dtypes
```

```python
df_processed["Date"] = pd.to_datetime(df_processed["Date"])

df_processed.info()
```

```python
df_processed["Year"] = df_processed["Date"].dt.year
df_processed["Month"] = df_processed["Date"].dt.month
df_processed["Day"] = df_processed["Date"].dt.day
df_processed["DayOfWeek"] = df_processed["Date"].dt.day_name()
```

```python
df_processed.head()
```

```python
df_processed.drop("Date", axis=1, inplace=True)
```
